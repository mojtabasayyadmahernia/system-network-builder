```python
from src.traversal import available_choices

selected = set()
for pick in ["material", "effective", "transformative", "non-circumstantial-clause"]:
    open_systems = [s.id for s in available_choices(net, selected)]
    print(f"Still to decide: {open_systems}")
    print(f"  -> choosing '{pick}'\n")
    selected.add(pick)

print("Final selection:", sorted(selected))
print("Nothing left open:", [s.id for s in available_choices(net, selected)])
```

```python
from src.model import SelectionExpression
from src.validator import validate

bad = [
    ["material"],                                  # incomplete
    ["transformative", "effective"],               # entry unsatisfied
    ["relational", "attributive", "identifying"],  # mutual exclusivity
]
for features in bad:
    r = validate(net, SelectionExpression(rank="clause", features=set(features)))
    print(f"\n{features}")
    for e in r.errors:
        print(f"  [{e.rule.value}] {e.message}")
```
